### Advantages of PyTorch Dataset and DataLoader:

* A standardized interface for data. Regardless of what the data is (be it csv, image, text) PyTorch Dataset gives us a standardized way for data-access so that our training loop does not become dataset-specific.
* An easy way to apply transformations. Maybe we want to apply some transformations, like converting images to tensors, normalise, etc. Dataset class allows this.
* Shuffling and sampling. We can shuffle (which data goes into which batch).
* batch preparation and parallelization. 

- Think of it like this: Dataset allows to acess a single training sample (like image, label), whereas DataLoader allows us to split our data into samples, apply shuffling.
